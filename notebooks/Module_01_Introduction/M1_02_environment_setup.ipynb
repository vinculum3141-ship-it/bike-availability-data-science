## 📦 Part 2: Verify Package Installation

Let's verify that all required packages are installed and working correctly.

### Core Data Science Libraries

Run the cell below to check essential packages:

````python
# ═══════════════════════════════════════════════════════════
# 1. VERIFY CORE PACKAGES
# ═══════════════════════════════════════════════════════════

print("Checking core data science packages...\n")

packages_to_check = {
    'numpy': 'numpy',
    'pandas': 'pandas',
    'matplotlib': 'matplotlib',
    'seaborn': 'seaborn',
    'scikit-learn': 'sklearn',
    'requests': 'requests',
}

import importlib

all_installed = True
for package_name, import_name in packages_to_check.items():
    try:
        module = importlib.import_module(import_name)
        version = getattr(module, '__version__', 'unknown')
        print(f"✅ {package_name:20s} version {version}")
    except ImportError:
        print(f"❌ {package_name:20s} NOT INSTALLED")
        all_installed = False

print("\n" + "=" * 50)
if all_installed:
    print("🎉 All core packages installed successfully!")
else:
    print("⚠️  Some packages are missing. Please install them:")
    print("   pip install -r requirements.txt")
print("=" * 50)
````

import pandas as pd
from packaging import version

def check_pandas_version(min_version="2.0.0"):
    current = pd.__version__
    meets_req = version.parse(current) >= version.parse(min_version)
    
    if meets_req:
        print(f"✅ Pandas {current} meets requirement >= {min_version}")
    else:
        print(f"⚠️ Pandas {current} is below requirement >= {min_version}")
    
    return meets_req

import sys
import platform
import pandas as pd
import numpy as np
import matplotlib
import seaborn as sns
import sklearn

def create_environment_report(filename="environment_report.txt"):
    with open(filename, 'w') as f:
        f.write("=" * 50 + "\n")
        f.write("ENVIRONMENT REPORT\n")
        f.write("=" * 50 + "\n\n")
        
        f.write(f"Python Version: {sys.version.split()[0]}\n")
        f.write(f"OS: {platform.system()} {platform.release()}\n\n")
        
        f.write("Package Versions:\n")
        f.write("-" * 50 + "\n")
        packages = {
            'pandas': pd.__version__,
            'numpy': np.__version__,
            'matplotlib': matplotlib.__version__,
            'seaborn': sns.__version__,
            'scikit-learn': sklearn.__version__
        }
        
        for name, ver in packages.items():
            f.write(f"{name:20s}: {ver}\n")
        
        f.write(f"\nTotal packages checked: {len(packages)}\n")
    
    print(f"✅ Report saved to {filename}")

create_environment_report()
