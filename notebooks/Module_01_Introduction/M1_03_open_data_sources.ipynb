## 📖 Part 1: What is Open Data?

### Definition

**Open data** is data that is:
- ✅ **Freely available** - Anyone can access it
- ✅ **Machine-readable** - Structured format (CSV, JSON, XML)
- ✅ **Licensed for reuse** - Legal to use, modify, and share
- ✅ **Non-discriminatory** - Available to everyone
- ✅ **No registration required** (in many cases)

### Why Open Data Matters

**For Learning & Research:**
- 🎓 Practice with real-world data
- 📊 Build portfolio projects
- 🔬 Conduct research without cost barriers

**For Society:**
- 🏛️ Government transparency
- 💡 Innovation and entrepreneurship
- 👥 Citizen engagement
- 📈 Data-driven decision making

**For This Project:**
- 🚴 Real bike-sharing data from cities
- 🌦️ Real weather data from meteorological institutes
- 💰 **Free** - no API keys or payments required (mostly!)
- 🔄 **Reproducible** - anyone can run the same analysis

---

### Open Data Principles (The 5 Stars ⭐)

1. ⭐ **Available online** with open license
2. ⭐⭐ **Structured data** (e.g., Excel instead of PDF)
3. ⭐⭐⭐ **Non-proprietary format** (e.g., CSV instead of Excel)
4. ⭐⭐⭐⭐ **Uses URIs** to identify things
5. ⭐⭐⭐⭐⭐ **Linked data** - connected to other data

**Goal**: We aim for at least 3-star data in this project!

---

### Legal Considerations

**Always check:**
- 📜 **License** - Can you use it for education? Commercial use?
- 🔄 **Attribution** - Do you need to cite the source?
- 📊 **Terms of Service** - Any usage restrictions?
- 🚫 **Rate Limits** - How many requests per hour/day?

**Best Practice**:
```markdown
Data Source: [Name]
URL: [link]
License: [CC-BY-4.0, Open Database License, etc.]
Access Date: [YYYY-MM-DD]
```

### 🧪 Demo: Fetching Bike Data

Let's try fetching real bike-sharing data!

````python
# ═══════════════════════════════════════════════════════════
# DEMO: FETCH BIKE DATA FROM CITYBIKES API
# ═══════════════════════════════════════════════════════════

print("🚴 Fetching bike-sharing data from CityBikes API...\n")
print("🚴 Fetching OV-fiets data and filtering for Amsterdam stations...")

# OV-fiets API endpoint
url = "http://api.citybik.es/v2/networks/ov-fiets"

try:
    # Make API request
    response = requests.get(url, timeout=10)
    response.raise_for_status()  # Raise error for bad status codes

    # Parse JSON response
    data = response.json()

    # Extract network info
    network_info = data['network']
    print(f"✅ Successfully fetched data!")
    print(f"\n📊 Network Information:")
    print(f"   Name: {network_info['name']}")
    print(f"   City: {network_info['location']['city']}")
    print(f"   Country: {network_info['location']['country']}")
    print(f"   Total Stations: {len(network_info['stations'])}")

    # Filter for stations in Amsterdam
    amsterdam_ov_stations = []

    if network_info.get('stations'):
        for station in network_info['stations']:
            if 'name' in station and 'Amsterdam' in station['name']:
                amsterdam_ov_stations.append(station)

        if amsterdam_ov_stations:
            print(f"✅ Found {len(amsterdam_ov_stations)} OV-fiets station(s) with 'Amsterdam' in their name.")
            for station in amsterdam_ov_stations:
                print(f"   - Station: {station['name']} (ID: {station['id']})")
                print(f"     Location: ({station['latitude']:.4f}, {station['longitude']:.4f})")
                print(f"     Bikes available: {station['free_bikes']}")
                print(f"     Docks available: {station['empty_slots']}")
                print(f"     Timestamp: {station['timestamp']}")
                print()

        else:
            print("No OV-fiets stations found with 'Amsterdam' in their name.")
            print("Note: The 'OV-fiets' network covers all of the Netherlands, not just Amsterdam.")
    else:
        print("❌ No station data found in the OV-fiets network.")

    print("=" * 60)
    print("🎉 API demo successful! We can fetch real-time bike data!")
    print("=" * 60)

except requests.exceptions.RequestException as e:
    print(f"❌ Error fetching OV-fiets data: {e}")
    print("   Skipping data fetching for now.")
````

### 📊 Converting to DataFrame

Let's convert this JSON data into a pandas DataFrame:

````python
# ═══════════════════════════════════════════════════════════
# CONVERT BIKE DATA TO DATAFRAME
# ═══════════════════════════════════════════════════════════

try:
    # Extract stations data
    stations = amsterdam_ov_stations

    # Create DataFrame
    bike_df = pd.DataFrame([
        {
            'station_id': s['id'],
            'station_name': s['name'],
            'latitude': s['latitude'],
            'longitude': s['longitude'],
            'bikes_available': s['free_bikes'],
            'docks_available': s['empty_slots'],
            'timestamp': pd.to_datetime(s['timestamp'].replace('Z', '')) # Remove 'Z' for parsing
        }
        for s in stations
    ])

    print("📊 Bike Data DataFrame:\n")
    print(bike_df.head())

    print(f"\n📈 Quick Statistics:")
    print(f"   Total stations: {len(bike_df)}")
    print(f"   Total bikes available: {bike_df['bikes_available'].sum()}")
    print(f"   Average bikes per station: {bike_df['bikes_available'].mean():.1f}")
    print(f"   Stations with no bikes: {(bike_df['bikes_available'] == 0).sum()}")

except Exception as e:
    print(f"Note: Could not create DataFrame. This is okay for the demo!")
    print(f"We'll work with pre-downloaded data in later notebooks.")
````

### 🧪 Demo: Fetching Weather Data

Let's fetch real weather data!

````python
# ═══════════════════════════════════════════════════════════
# DEMO: FETCH WEATHER DATA FROM OPEN-METEO
# ═══════════════════════════════════════════════════════════

print("🌦️ Fetching weather data from Open-Meteo API...\n")

# Amsterdam coordinates
latitude = 52.3676
longitude = 4.9041

# Open-Meteo API endpoint (current weather)
url = f"https://api.open-meteo.com/v1/forecast"
params = {
    'latitude': latitude,
    'longitude': longitude,
    'current_weather': True,
    'timezone': 'Europe/Amsterdam'
}

try:
    # Make API request
    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    # Parse JSON response
    weather_data = response.json()
    current = weather_data['current_weather']

    print(f"✅ Successfully fetched weather data!")
    print(f"\n🌡️ Current Weather in Amsterdam:")
    print(f"   Temperature: {current['temperature']}°C")
    print(f"   Wind Speed: {current['windspeed']} km/h")
    print(f"   Wind Direction: {current['winddirection']}°")
    print(f"   Time: {current['time']}")

    # Weather code interpretation
    weather_codes = {
        0: "Clear sky ☀️",
        1: "Mainly clear 🌤️",
        2: "Partly cloudy ⛅",
        3: "Overcast ☁️",
        45: "Foggy 🌫️",
        48: "Foggy 🌫️",
        51: "Light drizzle 🌦️",
        61: "Light rain 🌧️",
        80: "Rain showers 🌧️",
        95: "Thunderstorm ⛈️"
    }

    weather_desc = weather_codes.get(current['weathercode'], "Unknown")
    print(f"   Conditions: {weather_desc}")

    print("\n" + "=" * 60)
    print("🎉 Weather API demo successful!")
    print("=" * 60)

except requests.exceptions.RequestException as e:
    print(f"❌ Error fetching data: {e}")
    print("   Don't worry! We'll work with sample data instead.")
````

### 📊 Weather Data for Time Series

For our prediction model, we'll need **historical** weather data:

````python
# ═══════════════════════════════════════════════════════════
# DEMO: FETCH HISTORICAL WEATHER DATA
# ═══════════════════════════════════════════════════════════

print("📅 Fetching historical weather data (past week)...\n")

from datetime import datetime, timedelta

# Date range (past 7 days)
end_date = datetime.now()
start_date = end_date - timedelta(days=7)

# Open-Meteo API endpoint (historical)
url = "https://api.open-meteo.com/v1/forecast"
params = {
    'latitude': latitude,
    'longitude': longitude,
    'start_date': start_date.strftime('%Y-%m-%d'),
    'end_date': end_date.strftime('%Y-%m-%d'),
    'hourly': 'temperature_2m,precipitation,windspeed_10m',
    'timezone': 'Europe/Amsterdam'
}

try:
    response = requests.get(url, params=params, timeout=10)
    response.raise_for_status()

    weather_history = response.json()
    hourly = weather_history['hourly']

    # Create DataFrame
    weather_df = pd.DataFrame({
        'timestamp': pd.to_datetime(hourly['time']),
        'temperature': hourly['temperature_2m'],
        'precipitation': hourly['precipitation'],
        'windspeed': hourly['windspeed_10m']
    })

    print("📊 Historical Weather DataFrame:\n")
    print(weather_df.head(10))

    print(f"\n📈 Quick Statistics (past week):")
    print(f"   Average temperature: {weather_df['temperature'].mean():.1f}°C")
    print(f"   Max temperature: {weather_df['temperature'].max():.1f}°C")
    print(f"   Min temperature: {weather_df['temperature'].min():.1f}°C")
    print(f"   Total precipitation: {weather_df['precipitation'].sum():.1f} mm")
    print(f"   Average wind speed: {weather_df['windspeed'].mean():.1f} km/h")

    print("\n✅ Historical weather data successfully retrieved!")

except Exception as e:
    print(f"Note: Could not fetch historical data.")
    print("We'll work with pre-downloaded data in later notebooks.")
````
