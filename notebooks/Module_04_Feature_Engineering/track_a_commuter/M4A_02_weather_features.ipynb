## 2. Temperature Features

Create temperature-based features that capture comfort zones.

```python
# Assume temperature column exists (adjust column name as needed)
# Common weather API fields: 'temperature', 'temp', 'temp_c'

# Temperature binning (Celsius)
def create_temp_bins(temp):
    """
    Bin temperature into comfort categories.
    Ranges based on typical cycling comfort:
    - Very Cold: < 5°C
    - Cold: 5-10°C
    - Cool: 10-15°C
    - Comfortable: 15-25°C (optimal)
    - Warm: 25-30°C
    - Hot: > 30°C
    """
    if temp < 5:
        return 'very_cold'
    elif temp < 10:
        return 'cold'
    elif temp < 15:
        return 'cool'
    elif temp < 25:
        return 'comfortable'
    elif temp < 30:
        return 'warm'
    else:
        return 'hot'

# Example: Apply to temperature column (adjust column name)
# df['temp_category'] = df['temperature'].apply(create_temp_bins)

# Binary indicators for extreme temperatures
# df['is_very_cold'] = (df['temperature'] < 5).astype(int)
# df['is_very_hot'] = (df['temperature'] > 30).astype(int)
# df['is_comfortable_temp'] = ((df['temperature'] >= 15) & (df['temperature'] <= 25)).astype(int)

print("Temperature features created")
# print(df[['temperature', 'temp_category', 'is_comfortable_temp']].describe())
```
